## Eval function

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    mean_squared_error
)

def eval_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    # Regression-style metrics
    mse = mean_squared_error(y_test, y_pred)
    rmse = mse ** 0.5

    # Classification metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_prob)

    print(f"MSE       : {mse:.4f}")
    print(f"RMSE      : {rmse:.4f}")
    print(f"Accuracy  : {accuracy:.4f}")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1 Score  : {f1:.4f}")
    print(f"ROC-AUC   : {roc_auc:.4f}")

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Numerical columns to standardize
num_cols = [
    'age',
    'fnlwgt',
    'education.num',
    'capital.gain',
    'capital.loss',
    'hours.per.week'
]

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols)
    ],
    remainder='passthrough'
)

# Logistic Regression pipeline
logistic_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])

# Train
logistic_model.fit(X_train, y_train)
```

```python
eval_model(logistic_model, X_test, y_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', KNeighborsClassifier(n_neighbors=5))
])

knn_model.fit(X_train, y_train)

print("KNN")
eval_model(knn_model, X_test, y_test)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest")
eval_model(rf_model, X_test, y_test)
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    eval_metric='logloss'
)

xgb_model.fit(X_train, y_train)

print("XGBoost")
eval_model(xgb_model, X_test, y_test)
```

```python
from sklearn.svm import SVC

svm_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', SVC(
        kernel='rbf',
        probability=True,
        random_state=42
    ))
])

svm_model.fit(X_train, y_train)

print("SVM")
eval_model(svm_model, X_test, y_test)
```

```python
def eval_model(model, X_test, y_test, print_results=True):

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    mse = mean_squared_error(y_test, y_pred)
    rmse = mse ** 0.5

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_prob)

    if print_results:
        print(f"MSE       : {mse:.4f}")
        print(f"RMSE      : {rmse:.4f}")
        print(f"Accuracy  : {accuracy:.4f}")
        print(f"Precision : {precision:.4f}")
        print(f"Recall    : {recall:.4f}")
        print(f"F1 Score  : {f1:.4f}")
        print(f"ROC-AUC   : {roc_auc:.4f}")

        print("\nConfusion Matrix:")
        print(confusion_matrix(y_test, y_pred))

        print("\nClassification Report:")
        print(classification_report(y_test, y_pred))

    return {
        'MSE': mse,
        'RMSE': rmse,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1': f1,
        'ROC-AUC': roc_auc
    }
```

```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from scipy.stats import randint

rf = RandomForestClassifier(
    random_state=42,
    n_jobs=-1
)

rf_params = {
    'n_estimators': randint(100, 500),
    'max_depth': [None, 10, 20, 30, 40],
    'min_samples_split': randint(2, 15),
    'min_samples_leaf': randint(1, 8),
    'max_features': ['sqrt', 'log2', None],
    'class_weight': [None, 'balanced']
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

rf_search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=rf_params,
    n_iter=30,
    scoring='f1',
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_search.fit(X_train, y_train)
```

```python
print("Best RF parameters:")
print(rf_search.best_params_)

print("\nBest CV F1:")
print(rf_search.best_score_)
```

```python
best_rf = rf_search.best_estimator_

print("Tuned Random Forest")
rf_results = eval_model(best_rf, X_test, y_test)
```

```python
from xgboost import XGBClassifier
from scipy.stats import randint, uniform

xgb = XGBClassifier(
    objective='binary:logistic',
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

xgb_params = {
    'n_estimators': randint(100, 500),
    'max_depth': randint(3, 10),
    'learning_rate': uniform(0.01, 0.2),
    'subsample': uniform(0.6, 0.4),
    'colsample_bytree': uniform(0.6, 0.4),
    'min_child_weight': randint(1, 10),
    'gamma': uniform(0, 0.5)
}

xgb_search = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=xgb_params,
    n_iter=30,
    scoring='f1',
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

xgb_search.fit(X_train, y_train)
```

```python
best_xgb = xgb_search.best_estimator_

print("Tuned XGBoost")
xgb_results = eval_model(best_xgb, X_test, y_test)
```

```python
from sklearn.ensemble import VotingClassifier, StackingClassifier
from sklearn.linear_model import LogisticRegression

# Best version of each model family
ensemble_models = [
    ('lr', logistic_model),
    ('knn', knn_model),
    ('rf', best_rf),
    ('xgb', best_xgb),
    ('svm', svm_model)
]
```

```python
weighted_voting = VotingClassifier(
    estimators=ensemble_models,
    voting='soft',
    weights=[1, 1, 2, 3, 2],
    n_jobs=-1
)

weighted_voting.fit(X_train, y_train)

print("Weighted Soft Voting")
voting_results = eval_model(
    weighted_voting,
    X_test,
    y_test
)
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression

stack_models = [
    ('lr', logistic_model),
    ('rf', best_rf),
    ('xgb', best_xgb)
]

stacking_model = StackingClassifier(
    estimators=stack_models,
    final_estimator=LogisticRegression(max_iter=1000),
    cv=3,
    stack_method='predict_proba',
    n_jobs=-1
)

stacking_model.fit(X_train, y_train)

stacking_results = eval_model(
    stacking_model,
    X_test,
    y_test,
    print_results=True
)
```
